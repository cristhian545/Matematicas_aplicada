Determinar el modelo lineal que por los puntos $A$ (1,6) y $B$(3,10)

```python
import numpy as np
x=[1,3]
y=[6,10]

pendiente, intercepto = np.polyfit(x,y,1)
print(f"El modelo es f(x)={pendiente:.1f}x+{intercepto:1f}")
```
